```sql
CREATE SCHEMA IF NOT EXISTS trial.secure;

CREATE OR REPLACE VIEW trial.secure.v_hourly_device_health AS
SELECT a.*
FROM trial.gold.v_hourly_device_health a
WHERE EXISTS (
  SELECT 1
  FROM security.group_tenant_access m
  WHERE upper(m.tenant_id) = upper(a.TenantId)
    AND is_account_group_member(m.group_name)
);
```
